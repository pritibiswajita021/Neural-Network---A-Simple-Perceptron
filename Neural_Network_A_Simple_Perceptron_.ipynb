Question 3: Describe the purpose of activation function in neural networks. Compare Sigmoid, ReLU, and Tanh functions.

Answer:

Purpose:

Introduces non-linearity
Enables complex pattern learning
| Function | Formula                   | Range  | Pros            | Cons                     |
| -------- | ------------------------- | ------ | --------------- | ------------------------ |
| Sigmoid  | 1/(1+e^-x)                | (0,1)  | Smooth          | Vanishing gradient       |
| Tanh     | (e^x - e^-x)/(e^x + e^-x) | (-1,1) | Zero-centered   | Still vanishing gradient |
| ReLU     | max(0,x)                  | [0,∞)  | Fast, efficient | Dead neurons             |

Question 4: What is the difference between Loss function and Cost function in neural networks? Provide examples.

Answer:

Loss Function: Error for a single data point
Cost Function: Average loss over dataset

Examples:

Loss: Mean Squared Error (MSE), Cross-Entropy
Cost: Average of MSE over all samples

```python
'''Question 6: Write a Python program to implement a single-layer perceptron using NumPy for AND gate.'''
import numpy as np

# Input and output
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([0,0,0,1])

# Initialize weights and bias
weights = np.random.rand(2)
bias = np.random.rand()

# Activation function
def step(x):
    return 1 if x >= 0 else 0

# Training
lr = 0.1
for epoch in range(10):
    for i in range(len(X)):
        linear = np.dot(X[i], weights) + bias
        y_pred = step(linear)
        error = y[i] - y_pred

        weights += lr * error * X[i]
        bias += lr * error

# Testing
for i in range(len(X)):
    print(X[i], "->", step(np.dot(X[i], weights) + bias))
```

```python
'''Question 7: Implement and visualize Sigmoid, ReLU, and Tanh activation functions'''
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10,10,100)

sigmoid = 1/(1+np.exp(-x))
relu = np.maximum(0,x)
tanh = np.tanh(x)

plt.plot(x, sigmoid, label="Sigmoid")
plt.plot(x, relu, label="ReLU")
plt.plot(x, tanh, label="Tanh")
plt.legend()
plt.title("Activation Functions")
plt.show()
```

```python
'''Question 8: Use Keras to build and train a simple multilayer neural network on the
MNIST digits dataset. Print the training accuracy'''
import tensorflow as tf
from tensorflow import keras

(x_train, y_train), _ = keras.datasets.mnist.load_data()
x_train = x_train / 255.0

model = keras.Sequential([
    keras.layers.Flatten(),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

history = model.fit(x_train, y_train, epochs=5)

print("Training Accuracy:", history.history['accuracy'][-1])
```

```python
'''Question 9: Visualize the loss and accuracy curves for a neural network model trained
on the Fashion MNIST dataset. Interpret the training behavior.'''
from tensorflow import keras
import matplotlib.pyplot as plt

(x_train, y_train), _ = keras.datasets.fashion_mnist.load_data()
x_train = x_train / 255.0

model = keras.Sequential([
    keras.layers.Flatten(),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

history = model.fit(x_train, y_train, epochs=5)

# Plot
plt.plot(history.history['loss'], label='Loss')
plt.plot(history.history['accuracy'], label='Accuracy')
plt.legend()
plt.title("Training Curves")
plt.show()

'''Interpretation:

Loss decreases → model learning
Accuracy increases → improving performance
Overfitting if gap appears (not here)'''
```
